#### **5)** The code cell below generates a random integer between 1 and 20 (both 1 and 20 are possible values). 

#### Write code that tests the following:

- #### Tests if the number satisfies $1 \leq n < 5$.

- #### Tests if the number satisfies $5 \leq n \leq 10$

- #### Test if the number satisfies $10 < n < 15$

#### If it satisfies one of the conditions, it prints a message saying which one. If it doesn't satisfy any of the conditions, print a statement saying the number is greater than or equal to 15.

```python
import random

lo = 1
hi = 20

random_int = random.randint(lo, hi)
print(random_int)

# SOLUTION

if (1 <= random_int) and (random_int < 5):
    print("In the interval [1, 5)") 
elif (5 <= random_int) and (random_int <= 10):
    print("In the interval [5, 10]")
elif (10 < random_int) and (random_int < 15):
    print("In the interval (10, 15)")
else:
    print("Greater than or equal to 15")
```

#### **8)** Run the cell below. How do you explain the output you see?

```python
list_1 = [0, 0, 0, 0, 0]
list_2 = [5, 50, 75, 100, -0.01]
print(list_1 <= list_2)

# Explanation: This is very similar to alphabetical ordering

# If we put the words "Zebra" and "Apple" in alphabetical order, we
# would put Apple before Zebra, because A comes before Z in the alphabet.
# The endings "pple" and "ebra" do not need to be considered.

# Lists work in a similar way

# Above, since list_1[0] = 0, list_2[0] = 5, and "0 <= 5" is True, 
# Python declares that the statement list_1 <= list_2 is True

# So even though list_1[4] <= list_2[4] is False, 
# comparing the entire lists evaluates to True
```

#### **9)** Predict what the output of the cell below will be **before** running it. What is happening in the code?

```python
x = 7
y = 3

z = x*y + 9

q = z/10 

x = x - q 

x = x == 4

# Predict what this will print, then uncomment and run the cell to check your answer
print(x) 

# See next cell for explanation
```

```python
# We will keep track of the values of each variable

# We assign 7 to "x" and assign 3 to "y"
x = 7 
y = 3

# x: 7
# y: 3

# compute 7*3 + 9 = 30, and assign it to "z"
z = x*y + 9 

# z: 30

# Compute 30/10 = 3 and assign it to "q"
q = z/10 

# q: 3

# Compute 7 - 3 = 4 and assign it to "x", overwriting the old value of 7
x = x - q 

# x: 7̶ 4

# Compare the variable x with the integer 4, using "=="
# This evaluates to True
# Then assign True to "x", overwriting the value of 4

x = x == 4

# x: 7̶ 4̶ True

print(x) # Since x is True, this will print True
```

#### **11)** How should we check whether `x-z` and `y` are "equal"? Code up your approach below

```python
x = 1
y = 0.1
z = 0.9

# SOLUTION 1
print("x - z and y are equal:", abs(x - z - y) < 1e-13)

# SOLUTION 2
t = x - z
print("x - z and y are equal:", abs(t - y) < 1e-13)

# SOLUTION 3
t = x - z
diff = t - y
is_equal = abs(diff) < 1e-13

print("x - z and y are equal:", is_equal)
```
